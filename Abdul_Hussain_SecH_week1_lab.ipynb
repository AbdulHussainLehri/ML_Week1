```python
# Section A
# Q1
a = 10
b = 20

print("Before swapping:")
print("a =", a)
print("b =", b)

a, b = b, a

print("\nAfter swapping:")
print("a =", a)
print("b =", b)

# Q2
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True

print(is_prime(17))
print(is_prime(20))

# Q3
n = 10
a = 0
b = 1

print("Fibonacci sequence:")
for i in range(n):
    print(a, end=" ")
    a, b = b, a + b

# Q4
def remove_duplicates(numbers):
    result = []

    for num in numbers:
        if num not in result:
            result.append(num)

    return result


numbers = [1, 2, 2, 3, 4, 3, 5, 1]

print(remove_duplicates(numbers))

# Q5
def multiply(*args):
    result = 1

    for num in args:
        result *= num

    return result
print(multiply(2, 3))
print(multiply(2, 3, 4))
print(multiply(5, 2, 3))

# Q6

text = "banana"

frequency = {
    char: text.count(char)
    for char in set(text)
}

print(frequency)

# Q7

employees = [
    {"name": "Ali", "department": "IT", "salary": 70000},
    {"name": "Sara", "department": "HR", "salary": 85000},
    {"name": "Ahmed", "department": "Finance", "salary": 75000},
    {"name": "Omar", "department": "IT", "salary": 95000}
]

highest_salary_employee = max(employees, key=lambda employee: employee["salary"])

print("Employee with highest salary:")
print(highest_salary_employee)

```
